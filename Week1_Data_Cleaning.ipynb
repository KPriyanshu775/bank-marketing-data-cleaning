```python
import pandas as pd
import numpy as np

df = pd.read_csv("bank-additional-full.csv", sep=";")

duplicates = df[df.duplicated()]

print("Dataset shape:", df.shape)
print("Number of duplicate rows:", len(duplicates))

duplicates
```

```python
df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Rows after removing duplicates:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
print("Cleaned dataset shape:", df_clean.shape)
```

```python
df_clean.describe()
```

```python
print("Zero-duration calls:",
      (df_clean["duration"] == 0).sum())

print("Records with pdays = 999:",
      (df_clean["pdays"] == 999).sum())

print("Campaign contacts greater than 10:",
      (df_clean["campaign"] > 10).sum())
```

```python
Q1 = df_clean["campaign"].quantile(0.25)
Q3 = df_clean["campaign"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = df_clean[
    (df_clean["campaign"] < lower_bound) |
    (df_clean["campaign"] > upper_bound)
]

print("Potential campaign outliers:", len(outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 4))

sns.boxplot(x=df_clean["campaign"])

plt.title("Box Plot of Campaign Contacts")
plt.xlabel("Number of Contacts")
plt.show()
```

```python
print("Minimum duration:", df_clean["duration"].min())
print("Median duration:", df_clean["duration"].median())
print("Maximum duration:", df_clean["duration"].max())
print("Zero-duration calls:", (df_clean["duration"] == 0).sum())
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df_clean, x="duration", bins=50)

plt.title("Distribution of Call Duration")
plt.xlabel("Call duration (seconds)")
plt.ylabel("Number of calls")
plt.show()
```

```python
categorical_columns = [
    "job", "marital", "education",
    "default", "housing", "loan",
    "contact", "poutcome", "y"
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df_clean[col].value_counts(dropna=False))
```
